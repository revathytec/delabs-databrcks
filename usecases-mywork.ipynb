## DBUtils tasks

```python
%sql
create volume if not exists workspace.default.dbutilsvol;
```

```python
base_path = "/Volumes/workspace/default/dbutilsvol"
raw_path = f"{base_path}/raw"
processed_path = f"{base_path}/processed"
archive_path = f"{base_path}/archive"
dbutils.fs.mkdirs(raw_path)
dbutils.fs.mkdirs(processed_path)
dbutils.fs.mkdirs(archive_path)
```

```python
sample_data ="""patient_id,patient_name,age,gender
1,John Doe,68,M
2,Jane Smith,54,F
"""
dbutils.fs.put(f"{raw_path}/patients.csv", sample_data, True)
display(dbutils.fs.ls(raw_path))
```

```python
dbutils.fs.cp(f"{raw_path}/patients.csv", f"{processed_path}/patients.csv")
```

```python
dbutils.widgets.dropdown("env", "dev", ["dev", "qa", "prod"], "which Environment?")
dbutils.widgets.text("ownername", "", "enter Owner Name")
```

```python
envir = dbutils.widgets.get("env")
ownr = dbutils.widgets.get("ownername")

print(f"Environment: {envir} | Owner: {ownr}")
```

```python
source_file = f"{processed_path}/patients.csv"
archive_file = f"{archive_path}/patients.csv"

# Move file from processed to archive
dbutils.fs.mv(source_file, archive_file)

print(f"Moved patients.csv to archive | environment={envir} | owner={ownr}")
```

```python
old_name = f"{archive_path}/patients.csv"
new_name = f"{archive_path}/patients_{envir}.csv"

dbutils.fs.mv(old_name, new_name)
```
